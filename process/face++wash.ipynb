```python
import requests
import json
import os
import openpyxl
import ssl
print(ssl.OPENSSL_VERSION)
os.environ['HTTP_PROXY'] = ''
os.environ['HTTPS_PROXY'] = ''
os.environ['ALL_PROXY'] = ''
#Face++图像分析

# Face++ API的key和secret
API_KEY = '4U9N2ipGiVggzPXx_6Udp6XfEZPKO3Iw'
API_SECRET = 'tCRjEoK7ToMJ8bCH1asw60RBn8x6n3C7'

# 请求URL
FACE_DETECT_URL = 'https://api-cn.faceplusplus.com/facepp/v3/detect'

# 上传图片路径
IMAGE_FOLDER = r'D:\Users\AAA\Desktop\TEST'  # 替换为图片所在文件夹的路径
OUTPUT_FOLDER = r'd:\Users\AAA\Desktop\TEST'  # 保存分析数据的文件夹

# 要检测的人脸信息
DETECT_ATTRIBUTES = 'gender,age,smiling,blur,headpose,eyestatus,emotion,facequality,beauty,skinstatus,mouthstatus,eyegaze,nose_occlusion,chin_occlusion,face_occlusion'

# 创建保存数据的输出文件夹
if not os.path.exists(OUTPUT_FOLDER):
    os.makedirs(OUTPUT_FOLDER)

# 获取图片文件夹中的所有文件名
image_files = [f for f in os.listdir(IMAGE_FOLDER) if os.path.isfile(os.path.join(IMAGE_FOLDER, f))]

# 创建Excel文件
excel_file = os.path.join(OUTPUT_FOLDER, 'SAE.xlsx')
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Face Analysis Data"

# 写入Excel表头
headers = ['Image Name', 'Face Rectangle', 'Landmarks', 'Gender', 'Age', 'Face Quality', 'Blur', 'Emotion', 'Beauty', 'Skin Status','Smile','Head Pose','Eye Status','Mouth Status','Eye Gaze','Nose Occlusion','Chin Occlusion','Face Occlusion']
ws.append(headers)

def detect_face(image_path):
    """上传图片到Face++ API并返回检测结果"""
    with open(image_path, 'rb') as img_file:
        files = {'image_file': img_file}
        data = {
            'api_key': API_KEY,
            'api_secret': API_SECRET,
            'return_attributes': DETECT_ATTRIBUTES,
            'return_landmark': 2  # 获取86个人脸关键点
        }
        response = requests.post(FACE_DETECT_URL, data=data, files=files)
        return response.json()


def save_analysis_data_to_excel(image_name, data):
    """将检测结果保存到Excel文件"""
    if 'faces' in data and len(data['faces']) > 0:
        face_data = data['faces'][0]

        # 提取需要的字段
        face_rectangle = face_data['face_rectangle']
        landmark = face_data['landmark']
        attributes = face_data['attributes']

        gender = attributes['gender']['value']
        age = attributes['age']['value']
        face_quality = attributes['facequality']['value']
        blur = attributes['blur']['blurness']['value']
        emotion = json.dumps(attributes.get('emotion', {}), ensure_ascii=False)  # 将情绪转换为JSON字符串
        beauty = json.dumps(attributes.get('beauty', {}), ensure_ascii=False)  # 将beauty转换为JSON字符串
        skin_status = json.dumps(attributes.get('skinstatus', {}), ensure_ascii=False)  # 将皮肤状态转换为JSON字符串
        smiling = json.dumps(attributes.get('smile', {}), ensure_ascii=False) 
        headpose = json.dumps(attributes.get('headpose', {}), ensure_ascii=False) 
        eyestatus = json.dumps(attributes.get('eyestatus', {}), ensure_ascii=False)
        mouthstatus = json.dumps(attributes.get('mouthstatus', {}), ensure_ascii=False)
        eyegaze = json.dumps(attributes.get('eyegaze', {}), ensure_ascii=False)
        nose_occlusion = attributes['nose_occlusion']['nose_occ']
        chin_occlusion = attributes['chin_occlusion']['chin_occ']
        face_occlusion = json.dumps(attributes.get('face_occlusion', {}), ensure_ascii=False)
        # 写入一行数据
        ws.append([
            image_name,
            json.dumps(face_rectangle, ensure_ascii=False),  # face_rectangle转为字符串保存
            json.dumps(landmark, ensure_ascii=False),  # landmark转为字符串保存
            gender,
            age,
            face_quality,
            blur,
            emotion,
            beauty,
            skin_status,
            smiling,
            headpose,
            eyestatus,
            mouthstatus,
            eyegaze,
            nose_occlusion,
            chin_occlusion,
            face_occlusion
        ])

# 遍历图片文件，分析并保存检测结果
for image_file in image_files:
    image_path = os.path.join(IMAGE_FOLDER, image_file)
    print(f'正在处理: {image_file}')
    try:
        analysis_data = detect_face(image_path)
        save_analysis_data_to_excel(image_file, analysis_data)
        print(f'分析数据已保存到Excel: {image_file}')
    except Exception as e:
        print(f'处理 {image_file} 时发生错误: {e}')

# 保存Excel文件
wb.save(excel_file)
print(f'所有图片处理完毕，分析数据已保存到{excel_file}')
```

```python
import pandas as pd
import json
import re
'''  
    新版openart数据处理
    邹子露
    2025.5.17
'''

# 文件路径列表（请根据实际路径修改）
file_paths = [
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_ar.xlsx',
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_zh.xlsx',
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_sw.xlsx',
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_es.xlsx',
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_id.xlsx',
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_hi.xlsx',
    r'd:\Users\AAA\Desktop\517\face++\firststep\open_SAE.xlsx'
]
initial_file_path = r'd:\Users\AAA\Desktop\517\face++\processed\initial.xlsx'
outputs_path = r'd:\Users\AAA\Desktop\517\face++\processed\Processed_Open.csv'

# 创建一个空的 DataFrame，用于存储所有处理后的数据
all_data = pd.DataFrame()

# 定义一个函数，使用更严格的正则表达式提取语言编码
def extract_language(image_name):
    # 正则表达式严格匹配两位字母的语言编码和后续的数字递进编号
    match = re.search(r'_([a-z]{2,3})\d', image_name, re.IGNORECASE)
    if match:
        return match.group(1)
    else:
        return None  # 如果匹配不到，返回None

# 处理递进步骤文件
for file_path in file_paths:
    # 读取每个文件的数据
    data = pd.read_excel(file_path)
    # 使用正则表达式提取语言编号和递进步骤编号
    data['Language'] = data['Image Name'].apply(extract_language)  # 使用新的正则表达式语言提取逻辑
    data['Group'] = data['Image Name'].apply(lambda x: x.split('_')[0][4:])  # 提取图片组编号
    #data['Step'] = data['Image Name'].apply(lambda x: int(x.split('_')[1][2]))  # 提取递进步骤编号（1、2、3）
    #data['Step'] = data['Image Name'].apply(lambda x: re.search(r'_(\w+)$', x).group(1) if re.search(r'_(\w+)$', x) else None)
    data['Step'] = data['Image Name'].apply(lambda x: int(re.search(r'_[a-zA-Z]{2,}(\d)', x).group(1)))

    # 解析Beauty列中的JSON数据，并提取male_score和female_score
    data['male_score'] = data['Beauty'].apply(lambda x: json.loads(x)['male_score'])
    data['female_score'] = data['Beauty'].apply(lambda x: json.loads(x)['female_score'])
    data['avg_beauty_score'] = (data['male_score'] + data['female_score']) / 2  # 计算平均分
    
    # 提取其他应变量数据列
    data['Face Quality'] = data['Face Quality']  # 假设列名为 'Face Quality'
    data['Age'] = data['Age']  # 假设列名为 'Age'
    data['Skin Status_health'] = data['Skin Status'].apply(lambda x: json.loads(x)['health'])
    data['Skin Status_stain'] = data['Skin Status'].apply(lambda x: json.loads(x)['stain'])
    data['Skin Status_dark_circle'] = data['Skin Status'].apply(lambda x: json.loads(x)['dark_circle'])
    data['Skin Status_acne'] = data['Skin Status'].apply(lambda x: json.loads(x)['acne'])
    
    # 将性别转为二元分类变量（0代表女性，1代表男性）
    data['Gender'] = data['Gender'].map({'Male': 1, 'Female': 0})
    
    
    valid_languages = ['SAE', 'sw', 'zh', 'ar', 'hi', 'id', 'es'] 
    data = pd.get_dummies(data, columns=['Language'], prefix='Language', prefix_sep='_')
    
    
    for lang in valid_languages:
        if f'Language_{lang}' not in data.columns:
            data[f'Language_{lang}'] = 0  # 如果某个语言列缺失，补0
    
    # 将处理后的数据添加到all_data
    all_data = pd.concat([all_data, data], ignore_index=True)

# 读取initial文件并处理，设置Step为0，保留语言信息
initial_data = pd.read_excel(initial_file_path)

# 使用正则表达式提取语言编号
initial_data['Language'] = initial_data['Image Name'].apply(extract_language)  # 使用新的正则表达式语言提取逻辑
initial_data['Group'] = initial_data['Image Name'].apply(lambda x: re.search(r'_(\d+)', x).group(1))  # 提取组编号
initial_data['Step'] = 0  # 设置为初始图像

# 解析Beauty列中的JSON数据，并提取male_score和female_score
initial_data['male_score'] = initial_data['Beauty'].apply(lambda x: json.loads(x)['male_score'])
initial_data['female_score'] = initial_data['Beauty'].apply(lambda x: json.loads(x)['female_score'])
initial_data['avg_beauty_score'] = (initial_data['male_score'] + initial_data['female_score']) / 2

# 提取其他应变量数据列
initial_data['Face Quality'] = initial_data['Face Quality']
initial_data['Age'] = initial_data['Age']
initial_data['Skin Status_health'] = initial_data['Skin Status'].apply(lambda x: json.loads(x)['health'])
initial_data['Skin Status_stain'] = initial_data['Skin Status'].apply(lambda x: json.loads(x)['stain'])
initial_data['Skin Status_dark_circle'] = initial_data['Skin Status'].apply(lambda x: json.loads(x)['dark_circle'])
initial_data['Skin Status_acne'] = initial_data['Skin Status'].apply(lambda x: json.loads(x)['acne'])

# 将性别转为二元分类变量（0代表女性，1代表男性）
initial_data['Gender'] = initial_data['Gender'].map({'Male': 1, 'Female': 0})

# 对语言进行独热编码，使用自定义的列名格式，如 Language_cn 等
initial_data = pd.get_dummies(initial_data, columns=['Language'], prefix='Language', prefix_sep='_')

# 保留7种语言的独热编码列
for lang in valid_languages:
    if f'Language_{lang}' not in initial_data.columns:
        initial_data[f'Language_{lang}'] = 0

# 将initial数据合并到all_data中
all_data = pd.concat([all_data, initial_data], ignore_index=True)

# 保存处理后的数据为一个新的CSV文件
all_data.to_csv(outputs_path, index=False)

# 打印前几行确认处理结果
print(all_data.head())
```

```python
import pandas as pd
import re
import os

# 文件路径
initial_path = r'D:\Users\AAA\Desktop\517\fairface_oasd\initial.xlsx'
file_path = r'd:\Users\AAA\Desktop\517\face++\processed\Processed_Open.csv'
# output_path = r'd:\Users\AAA\Desktop\517\face++\processed\Processed_Open_with_race.csv'

# 检查文件是否存在
if not os.path.exists(initial_path):
    raise FileNotFoundError(f"文件 {initial_path} 不存在！")
if not os.path.exists(file_path):
    raise FileNotFoundError(f"文件 {file_path} 不存在！")

# 种族与语言映射
race_language_mapping = {
    'Middle Eastern': 'ar',
    'White': 'SAE',
    'Black': 'sw',
    'East Asian': 'zh',
    'Indian': 'hi',
    'Southeast Asian': 'id',
    'Latino_Hispanic': 'es'
}

# 1. 读取initial.xlsx中的v_图像种族信息
try:
    initial_data = pd.read_excel(initial_path)
    print("initial.xlsx列名:", initial_data.columns.tolist())
    
    # 创建数字编号到race的映射字典
    v_race_dict = {}
    for _, row in initial_data.iterrows():
        if pd.notna(row.get('face_name_align', None)) and isinstance(row['face_name_align'], str):
            match = re.search(r'v_(\d+)', row['face_name_align'], re.IGNORECASE)
            if match:
                image_id = match.group(1)
                v_race_dict[image_id] = row.get('race', None)
    print(f"从initial.xlsx中提取到 {len(v_race_dict)} 条v_图像种族映射")
    print("示例映射:", dict(list(v_race_dict.items())[:3]))  # 打印前3条示例
except Exception as e:
    raise ValueError(f"读取initial.xlsx失败: {e}")

# 2. 读取Processed_Open.csv
try:
    data = pd.read_csv(file_path)
    print("\nProcessed_Open.csv前5行:")
    print(data.head())
    if 'Image Name' not in data.columns:
        raise KeyError("Processed_Open.csv中缺少 'Image Name' 列！")
except Exception as e:
    raise ValueError(f"读取Processed_Open.csv失败: {e}")

# 3. 添加race列
def get_race(image_name):
    if not isinstance(image_name, str):
        return None
    
    # 处理v_格式
    v_match = re.search(r'v_(\d+)', image_name, re.IGNORECASE)
    if v_match:
        image_id = v_match.group(1)
        return v_race_dict.get(image_id, None)
    
    # 处理open格式
    open_match = re.search(r'open(\d+)_([a-zA-Z]+)', image_name, re.IGNORECASE)
    if open_match:
        language_code = open_match.group(2)
        return next((k for k, v in race_language_mapping.items() if v == language_code), None)
    
    return None

print("\n正在添加race列...")
data['race'] = data['Image Name'].apply(get_race)

# 检查race列的分布
print("\nrace列分布统计:")
print(data['race'].value_counts(dropna=False))

# 4. 独热编码
print("\n正在进行独热编码...")
all_races = list(race_language_mapping.keys())
for race in all_races:
    data[f'race_{race}'] = (data['race'] == race).astype(int)

# 5. 保存结果
data.to_csv(file_path, index=False)
print(f"\n处理完成！结果已保存到 {file_path}")
print("\n最终数据预览（带race和独热编码）:")
print(data.head())

# 验证v_格式的数据是否处理正确
v_data = data[data['Image Name'].str.contains('v_', case=False, na=False)]
if not v_data.empty:
    print("\n验证v_格式数据处理结果:")
    print(v_data[['Image Name', 'race']].head())
else:
    print("\n警告：未找到v_格式的数据！")
```
